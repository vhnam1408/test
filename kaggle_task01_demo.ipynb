### 💡 Trả lời câu hỏi nghiên cứu: "Tại sao Kaggle cung cấp GPU miễn phí?"
Kaggle được Google chính thức mua lại vào ngày **08/03/2017**. Google duy trì chính sách cung cấp GPU miễn phí (NVIDIA Tesla T4, P100) trên Kaggle vì 4 lý do chiến lược:
1. **Onboarding hệ sinh thái Google Cloud (GCP & Vertex AI):** Kaggle thu hút hơn 15 triệu kỹ sư dữ liệu và AI. Khi các dự án nghiên cứu hoặc sản phẩm vượt ngưỡng hạn mức GPU miễn phí (30h/tuần), người dùng tự nhiên nâng cấp lên dịch vụ trả phí của GCP.
2. **Xây dựng cộng đồng AI toàn cầu & nguồn dữ liệu mở:** Kaggle là trung tâm chia sẻ datasets, pre-trained models, và benchmarks phục vụ trực tiếp cho các nghiên cứu AI mở.
3. **Quảng bá các framework & mô hình của Google:** Sân chơi lý tưởng để triển khai thực tế các công nghệ như Gemma, Gemini API, TensorFlow, JAX.
4. **Tối ưu hóa tài nguyên phần cứng nhàn rỗi (Spot / Preemptible Compute):** Khai thác hạ tầng GPU dư thừa tại các trung tâm dữ liệu toàn cầu của Google Cloud thay vì để lãng phí năng lượng.
